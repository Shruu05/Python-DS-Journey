```python
import numpy as np
import pandas as pd
```

```python
# set random seed for reproducibility
np.random.seed(123)
```

```python
# create dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples*class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
# create my dataset with imbalanced dataset
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0]*n_class_0
})
class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_1),
    'target': [1]*n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop = True)
```

```python
df.head()
```

```python
df['target'].value_counts()
```

## UpSampling

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
```

```python
df_minority_update = resample(
    df_minority, replace = True, # sample with replacement
    n_samples = len(df_majority),
    random_state = 42
)
```

```python
df_minority_update.shape
```

```python
df_minority_update.head()
```

```python
df_upsampled = pd.concat([df_majority, df_minority_update])
```

```python
df_upsampled['target'].value_counts()
```

## DownSampled

```python
df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
df_majority_update = resample(
    df_majority, replace = False, # sample with replacement
    n_samples = len(df_minority),
    random_state = 42
)
```

```python
df_majority_update.shape
```

```python
df_downsampled = pd.concat([df_minority, df_majority_update])
```

```python
df_downsampled['target'].value_counts()
```
