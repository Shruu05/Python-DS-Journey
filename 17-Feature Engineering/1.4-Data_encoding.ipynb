### Nomial OHE Encoding

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
# create a simple dataframe
df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'green', 'red', 'blue']
})
```

```python
df.head()
```

```python
# create on instance of OneHotEncoder
encoder = OneHotEncoder()
```

```python
# perform fit & transform
encoded = encoder.fit_transform(df[['color']]).toarray()
```

```python
encoder_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out())
encoder_df
```

```python
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df, encoder_df], axis=1)
```

### Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lbl_encoder = LabelEncoder()
```

```python
lbl_encoded = lbl_encoder.fit_transform(df[['color']])
lbl_encoded
```

```python
print(lbl_encoder.transform([['red']]))
print(lbl_encoder.transform([['blue']]))
print(lbl_encoder.transform([['green']]))
```

### Ordinal Encoding

```python
import pandas as pd
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
# create dataset for orinal encoding
df = pd.DataFrame({
    'size' : ['small', 'medium', 'large', 'medium', 'small', 'large']
})
```

```python
df
```

```python
encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
encoder.transform([['small']])
```

### Target gided ordinal encode

```python
import pandas as pd
```

```python
# create simple dataframe with categorical variable & target variable
df = pd.DataFrame({
    'city': ['NewYork','London', 'Paris','Tokyo','NewYork','Paris'],
    'price': [200,150,300,250,100,320]
})
```

```python
df
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
df[['price', 'city_encoded']]
```

#### Example on tips data

```python
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
import pandas as pd
```

```python
mean_bills = df.groupby('day')['total_bill'].mean().to_dict()
```

```python
mean_bills
```

```python
df['days_encoded'] = df['day'].map(mean_bills)
df[['total_bill', 'days_encoded']]
```
